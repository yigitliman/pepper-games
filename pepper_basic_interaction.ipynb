### File Transfer

```python
import paramiko

def download_file_from_pepper(host, port, username, password, remote_file_path, local_file_path):
    try:
        # Create an SSH client
        ssh = paramiko.SSHClient()
        ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())
        
        # Connect to the Pepper robot
        ssh.connect(host, port, username, password)
        
        # Use the SSH client to create an SFTP client
        sftp = ssh.open_sftp()
        
        # Download the file
        sftp.get(remote_file_path, local_file_path)
        print(f"Successfully downloaded {remote_file_path} to {local_file_path}")
        
        # Close the SFTP session and SSH connection
        sftp.close()
        ssh.close()
        
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage
host = '192.168.137.229'  # Replace with Pepper's IP address
port = 22
username = 'nao'  # Replace with Pepper's SSH username
password = 'CHANGE_ME'  # Replace with Pepper's SSH password (ask a lab member)
remote_file_path = '/home/nao/test.wav'
local_file_path = './test.wav'

download_file_from_pepper(host, port, username, password, remote_file_path, local_file_path)
```
